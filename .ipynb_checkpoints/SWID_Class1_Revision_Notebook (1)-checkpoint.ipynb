## Part 1: Variables and Data Types

A **variable** is a labelled box that stores a value. The name is the label. The single equals sign `=` means "put this value in the box". It does not mean "is equal to".

In class we stored information about a student. A line that starts with `#` is a **comment**. Python ignores comments, and they are notes for people.

```python
#Getting the student info
name = "Nchang Leonardine"
age = 12
gender = "female"
is_student = True
```

**Printing with an f string.** Put the letter `f` before the quote. Anything inside `{ }` is replaced by the value of that variable.

```python
print(f"The student with name, {name} is {age} years old")

# Output:
# The student with name, Nchang Leonardine is 12 years old
```

**The four basic data types**

| Type | What it holds | Example from class |
| --- | --- | --- |
| `str` | Text, always inside quotes | `"Nchang Leonardine"`, `"female"` |
| `int` | Whole numbers | `12` |
| `float` | Numbers with a decimal point | `1.73` |
| `bool` | `True` or `False` only | `True` |

Use `type()` to ask Python what type a value has. Print a variable to see its value.

```python
type(gender)

# Output:
# <class 'str'>
```

```python
print(name)

# Output:
# Nchang leonardine
```

```python
print(is_student)

# Output:
# True
```

**Rules for naming variables**

- Use letters, numbers and underscores only.
- Do not start with a number, so `2nd_place` is wrong and `place_2` is fine.
- No spaces. Write `first_name`, not `first name`.
- Capital letters matter, so `Age` and `age` are two different variables.
- Do not use Python keywords like `if`, `class` or `True` as names.
- Good habit: use lowercase words joined with underscores.

## Part 2: Arithmetic Operators

Python works like a calculator. We stored two numbers and tried every operator.

| Operator | Meaning |
| --- | --- |
| `+` | Add |
| `-` | Subtract |
| `*` | Multiply |
| `/` | Divide, always gives a float |
| `//` | Floor division, keeps only the whole part |
| `%` | Remainder after dividing |
| `**` | Power |

```python
a = 17
b = 2

#Arithmetic Operations

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)

# Output:
# 19
# 15
# 34
# 8.5
# 8
# 1
# 289
```

**Tip:** In a notebook, if the last line of a cell is just an expression, Colab shows its value without `print()`. That is why `a + b` below shows `19`. This only works for the last line in a cell.

**Markdown cells** (like this one) are for notes and headings. Use **+ Text** in Colab to add one. Use `#` at the start of a line for headings.

```python
a + b

# Output:
# 19
```

**Order of operations.** Python follows the maths rule: brackets first, then powers, then multiply and divide, then add and subtract.

```python
# Extra: order of operations
print(2 + 3 * 4)
print((2 + 3) * 4)

# Output:
# 14
# 20
```

## Part 3: Comparison Operators

Comparison operators ask a question. The answer is always `True` or `False`.

| Operator | Meaning |
| --- | --- |
| `==` | Equal to |
| `!=` | Not equal to |
| `>` and `<` | Greater than, less than |
| `>=` and `<=` | Greater than or equal to, less than or equal to |

Remember: `=` stores a value, and `==` asks a question.

```python
#Comparison Operators
print(a > b)
print(a < b)
print(a == b)
print(a != b)
print(a >= 17)
print(b <= 4)

# Output:
# True
# False
# False
# True
# True
# True
```

```python
#Is the primary school fee higher?
fee_for_primary > fee_for_secondary

# Output:
# False
```

## Part 4: Logical Operators

Logical operators join questions together.

| Operator | Meaning | Result |
| --- | --- | --- |
| `and` | Both sides must be True | True only if both are True |
| `or` | At least one side must be True | False only if both are False |
| `not` | Flips the answer | True becomes False |

```python
# Logical operators
# and, or, not
age = 29
has_id = True

print(age < 18 or has_id)

# Output:
# True
```

Why `True`? The first part `age < 18` is False, but the second part `has_id` is True, and `or` needs only one True.

More examples to try:

```python
# Extra: and, not
print(age >= 18 and has_id)
print(not has_id)

# Output:
# True
# False
```

## Part 5: Type Conversion

You can change a value from one type to another with `int()`, `float()` and `str()`.

**Important:** `int()` cuts off the decimal part. It does **not** round. So `1.73` becomes `1`, and `0.99` becomes `0`.

```python
height = 1.73
print(int(height))

# Output:
# 1
```

```python
val = .99
print(int(val))

# Output:
# 0
```

```python
print(int(0.99))

# Output:
# 0
```

Other conversions to remember:

```python
# Extra: other conversions
print(float(4))
print(str(25) + "5")
print(int("30") + 5)

# Output:
# 4.0
# 255
# 35
```

## Part 6: Strings, Join and Repeat

**Concatenation** means joining strings with `+`. Spaces are text too, so we add `" "` between words.

**Repeating** a string uses `*`.

```python
first_name = "Nchang"
last_name = "Leonardine"

#Concatenate...
print("Nchang"+" " + "Leonardine")

# Output:
# Nchang Leonardine
```

```python
print(first_name + " " + last_name)

# Output:
# Nchang Leonardine
```

```python
#Repeating (*)

print("Hi\t" * 10)

# Output:
# Hi	Hi	Hi	Hi	Hi	Hi	Hi	Hi	Hi	Hi	
```

**Special characters (escape characters)**

| Code | Meaning |
| --- | --- |
| `\n` | Moves to the next line |
| `\t` | Gives a tab space |

```python
# Extra: see \n in action
print("Line one\nLine two")

# Output:
# Line one
# Line two
```

**Careful:** `+` joins text with text and adds numbers with numbers. It cannot mix them. Use `str()` or an f string:

```python
# Extra: text plus a number
age = 24
print("I am " + str(age) + " years old")
print(f"I am {age} years old")

# Output:
# I am 24 years old
# I am 24 years old
```

## Part 7: Indexing

Every character in a string has a position called an **index**. Counting starts at **0**. Negative numbers count from the end.

| P | y | t | h | o | n |
| --- | --- | --- | --- | --- | --- |
| 0 | 1 | 2 | 3 | 4 | 5 |
| -6 | -5 | -4 | -3 | -2 | -1 |

```python
#Indexing
word = "Python"
print(word[0])

# Output:
# P
```

```python
# print letter t
print(word[2])

# Output:
# t
```

```python
#Printing letter with negative indexing
print(word[-4])

# Output:
# t
```

---
## Part 8: Slicing

Slicing takes a **part** of a string with `[start:stop]`. The start is included and the stop is **not** included. If you leave out the start, Python begins at the front. If you leave out the stop, Python goes to the end.

```python
#Slicing
print(word[0:3])

# Output:
# Pyt
```

```python
print(word[-4:])

# Output:
# thon
```

```python
# Extra: more slicing
print(word[:2])
print(word[2:])
print(word[::2])
print(word[::-1])
print(len(word))

# Output:
# Py
# thon
# Pto
# nohtyP
# 6
```

## Part 9: Working with Strings (String Methods)

A **method** is a built in tool that belongs to a value. We write it after a dot.

| Method | What it does |
| --- | --- |
| `.upper()` | Makes all letters capital |
| `.lower()` | Makes all letters small |
| `.replace(old, new)` | Swaps one piece of text for another |
| `in` | Checks if a piece of text is inside another. Gives True or False |

```python
# Working with strings
word = "data"
new_word = word.upper()
print(new_word)

# Output:
# DATA
```

```python
another_word = "BIgger"
print(another_word.lower())

# Output:
# bigger
```

```python
name = "Dane"
print(name.replace("D", "J"))

# Output:
# Jane
```

```python
name = "Pharel"
print("F" in name)

# Output:
# False
```

Why `False`? Python is **case sensitive**. `"F"` with a capital letter is not the same as `"f"`, and neither one is in `"Pharel"`.

Strings are not changed by methods. `.upper()` gives you a **new** string, so we saved it in `new_word`.

### Cheat sheet

| Topic | Remember |
| --- | --- |
| Variable | `name = value` stores a value |
| Types | `str`, `int`, `float`, `bool`. Check with `type()` |
| f string | `f"Hello {name}"` |
| Arithmetic | `+ - * / // % **` |
| Comparison | `== != > < >= <=` give True or False |
| Logical | `and`, `or`, `not` |
| `int()` | Cuts the decimal. It does not round |

| Slice | `word[start:stop]`. Stop is not included |
| Methods | `.upper()`, `.lower()`, `.replace()`, `in` |

### Common mistakes

| Mistake | Fix |
| --- | --- |
| `my name = 5` | No spaces: `my_name = 5` |
| `if age = 18` | Two equals signs to compare: `age == 18` |
| `"Age: " + 24` | `"Age: " + str(24)` or an f string |
| `print("Hello {name}")` | Add the `f`: `print(f"Hello {name}")` |
| `word[6]` on `"Python"` | The last index is 5, so use `word[5]` or `word[-1]` |
| Forgetting a quote or bracket | Check that every `(` and `"` is closed |

## 3. Operators

Operators let you calculate, compare and combine conditions.

### Arithmetic operators

````python
a = 17
b = 2

# Arithmetic operations
print(a + b)    # 19   add
print(a - b)    # 15   subtract
print(a * b)    # 34   multiply
print(a / b)    # 8.5  divide, always gives a float
print(a // b)   # 8    floor division, whole part only
print(a % b)    # 1    remainder
print(a ** 2)   # 289  power
````

### Comparison operators
These give `True` or `False`.

````python
# Comparison operators
# a = 17 and b = 2
print(a > b)     # True
print(a < b)     # False
print(a == b)    # False
print(a != b)    # True
print(a >= 17)   # True
print(b <= 4)    # True
````

**Important:** `=` stores a value in a variable. `==` compares two values. They are not the same.

````python
fee_for_primary = 300000
fee_for_secondary = 600000

# Is the primary school fee higher?
fee_for_primary > fee_for_secondary
````

````python
fee_for_secondary >= fee_for_primary
````

### The Visuals

````python
# Visual: comparing the two fees
fig, ax = plt.subplots(figsize=(6, 3.2))
bars = ax.bar(["Primary", "Secondary"], [fee_for_primary, fee_for_secondary], color=[NAVY, ORANGE])
ax.set_title("fee_for_primary > fee_for_secondary ?  False")
ax.set_ylabel("Fee")
for bar, v in zip(bars, [fee_for_primary, fee_for_secondary]):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 10000, f"{v:,}", ha="center", fontweight="bold")
plt.show()
````

### Logical operators
`and`, `or`, `not` combine conditions.

| Operator | Meaning | Example result |
|---|---|---|
| `and` | Both must be true | `True and False` gives `False` |
| `or` | At least one must be true | `True or False` gives `True` |
| `not` | Flips the answer | `not True` gives `False` |

````python
# Logical operators
age = 29
has_id = True

print(age < 18 or has_id)     # False or True gives True
print(age < 18 and has_id)    # False and True gives False
print(not has_id)
````

## 5. Strings

A string is text inside quotes.

### Joining and repeating
Joining strings together is called **concatenation**. Use `+`. Use `*` to repeat.

````python
first_name = "Nchang"
last_name = "Chenka"

# Concatenate
print("Nchang" + " " + "Chenka")
print(first_name + " " + last_name)

# Repeating (*)
print("Hi\t" * 10)
````

### Indexing
Each character has a position called an **index**. Counting starts at **0**.
You can also count from the end using **negative** numbers.

### Visual on indexing

````python
# Visual: indexing the word "Python"
word = "Python"
fig, ax = plt.subplots(figsize=(8, 2.6))
ax.axis("off")
for i, ch in enumerate(word):
    hi = (i == 2)
    ax.add_patch(patches.Rectangle((i, 0.8), 1, 1, fc=ORANGE if hi else LIGHT, ec=NAVY, lw=2))
    ax.text(i + 0.5, 1.3, ch, ha="center", va="center", fontsize=20,
            color="white" if hi else NAVY, fontweight="bold")
    ax.text(i + 0.5, 0.45, str(i), ha="center", color=NAVY, fontsize=12, fontweight="bold")
    ax.text(i + 0.5, 2.1, str(i - len(word)), ha="center", color="gray", fontsize=11)
ax.text(-0.3, 0.45, "index", ha="right", color=NAVY, fontsize=10)
ax.text(-0.3, 2.1, "negative", ha="right", color="gray", fontsize=10)
ax.set_xlim(-1.8, 6.2); ax.set_ylim(0, 2.6)
ax.set_title('word[2] and word[-4] are both "t"')
plt.show()
````

````python
word = "Python"
print(word[0])      # P

# print letter t
print(word[-3])

# Printing the letter with negative indexing
print(word[-4])
````

### Slicing
Slicing takes a **piece** of a string: `word[start:stop]`.
The `start` is included. The `stop` is **not** included.

### Visual on slicing

````python
# Visual: slicing
fig, axes = plt.subplots(2, 1, figsize=(8, 3.8))
for ax, (title, picked) in zip(axes, [("word[0:3]  gives  Pyt", {0, 1, 2}),
                                      ("word[-4:]  gives  thon", {2, 3, 4, 5})]):
    ax.axis("off")
    for i, ch in enumerate("Python"):
        ax.add_patch(patches.Rectangle((i, 0), 1, 1, fc=ORANGE if i in picked else LIGHT, ec=NAVY, lw=2))
        ax.text(i + 0.5, 0.5, ch, ha="center", va="center", fontsize=18,
                color="white" if i in picked else NAVY, fontweight="bold")
    ax.set_xlim(-0.2, 6.2); ax.set_ylim(-0.1, 1.5)
    ax.set_title(title, fontsize=11)
plt.tight_layout()
plt.show()
````

````python
print(word[0:3])    # Pyt
print(word[-4:])    # thon
````

## 6. Conditionals

Conditionals let your program **make decisions**. We use `if`, `elif` and `else`.

- `if` checks the first condition.
- `elif` (else if) checks another condition.
- `else` runs when nothing above was true.
- Remember the colon `:` and the **indentation** (spaces) on the next line.

````python
temperature = 20

if temperature > 25:
    print("Buy icecream")
else:
    print("Drink tea")
````

Change `temperature` to `28` and run it again. What happens?

### Many choices with `elif`
Grading rule: **A** is 80 and above. **B** is 70 to 79. **C** is 60 to 69.

### Conditionals

````python
# Visual: the decision path for the grade rule
fig, ax = plt.subplots(figsize=(9, 3.4))
ax.axis("off")
def box(x, y, w, h, text, color):
    ax.add_patch(patches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.05", fc=color, ec="none"))
    ax.text(x + w/2, y + h/2, text, ha="center", va="center", color="white", fontsize=9.5, fontweight="bold")
box(0, 1.3, 1.9, 0.8, "score >= 80 ?", NAVY)
box(2.7, 2.5, 1.5, 0.7, "A", ORANGE)
box(2.7, 1.3, 2.3, 0.8, "70 to 79 ?", NAVY)
box(5.8, 2.1, 1.3, 0.7, "B", ORANGE)
box(5.8, 0.7, 2.3, 0.8, "60 to 69 ?", NAVY)
box(8.9, 1.5, 1.0, 0.6, "C", ORANGE)
box(8.9, 0.2, 1.0, 0.6, "else", "#5B7DB1")
for (x1, y1, x2, y2, lab) in [(1.95, 1.9, 2.7, 2.8, "yes"), (1.95, 1.7, 2.7, 1.7, "no"),
                              (5.05, 1.9, 5.8, 2.4, "yes"), (5.05, 1.6, 5.8, 1.2, "no"),
                              (8.15, 1.3, 8.9, 1.8, "yes"), (8.15, 0.9, 8.9, 0.5, "no")]:
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1), arrowprops=dict(arrowstyle="->", color="gray", lw=1.5))
    ax.text((x1 + x2)/2, (y1 + y2)/2 + 0.12, lab, color="gray", fontsize=8.5)
ax.set_xlim(-0.2, 10.3); ax.set_ylim(0, 3.4)
ax.set_title("if / elif / else: Python checks from the top and stops at the first yes")
plt.show()
````

### The Code

````python
# For A 80+, B 70 to 79, C 60 to 69
score = 40

if score >= 80:
    print("You have an A")
elif (score >= 70 and score <= 79):
    print("You have a B")
elif (score >= 60 and score <= 69):
    print("You have a C")
else:
    print("Average! You can do better")

# 80 was "You have an A"
# 40 was "Average! You can do better"
````

## 7. Lists

A **list** stores many values in one variable. Lists use square brackets `[ ]`, and items are separated by commas.

Lists keep their order, and you **can change** them. Indexing works the same as in strings: it starts at 0, and negative numbers count from the end.

````python
fruits = ["mangoes", "apples", "oranges"]
print(fruits[0])
print(fruits[-1])
print(fruits[2])
````

### Changing a list

| Task | Code |
|---|---|
| Add one item to the end | `fruits.append("bananas")` |
| Remove an item by value | `fruits.remove("apples")` |
| Change an item by position | `fruits[0] = "berries"` |
| Count the items | `len(fruits)` |
| Add many items at the end | `fruits.extend(new_fruits)` |
| Delete a slice | `del fruits[1:3]` |

````python
fruits = ["mangoes", "apples", "oranges"]
history = [("Start", list(fruits))]     # we save each step so we can draw it later

fruits.append("bananas")
print(fruits)
history.append(('append("bananas")', list(fruits)))
````

````python
fruits.remove("apples")
print(fruits)
history.append(('remove("apples")', list(fruits)))
````

````python
fruits[0] = "berries"
print(fruits)
history.append(('fruits[0] = "berries"', list(fruits)))
````

````python
print(len(fruits))
````

````python
fruits.append("watermelon")
print(fruits)
print(len(fruits))
history.append(('append("watermelon")', list(fruits)))
````

````python
new_fruits = ["cherry", "date", "elderberry"]
fruits.extend(new_fruits)
print(fruits)
history.append(('extend(new_fruits)', list(fruits)))
````

````python
del fruits[1:3]
print(fruits)
history.append(('del fruits[1:3]', list(fruits)))
````

`append()` adds **one** item. `extend()` adds **every item** from another list.
`del fruits[1:3]` deletes positions 1 and 2. Position 3 stays, because the stop number is not included.

Here is the whole story in one picture.

### Visual: how the list changed at every step

````python
n = len(history)
fig, axes = plt.subplots(n, 1, figsize=(10, 1.0 * n + 0.6))
for ax, (step, items) in zip(axes, history):
    ax.axis("off")
    ax.text(-0.2, 0.5, step, ha="right", va="center", fontsize=9, family="monospace", color=NAVY)
    for i, it in enumerate(items):
        ax.add_patch(patches.Rectangle((i * 1.35, 0.1), 1.3, 0.8, fc=LIGHT if i % 2 == 0 else "#FBE3D0", ec=NAVY, lw=1.5))
        ax.text(i * 1.35 + 0.65, 0.5, it, ha="center", va="center", fontsize=8.5, color=NAVY)
    ax.set_xlim(-0.1, 9.8); ax.set_ylim(0, 1)
fig.suptitle("The fruits list after each step", color=NAVY, fontweight="bold")
plt.subplots_adjust(left=0.27, top=0.9)
plt.show()
````

## 8. Git and GitHub basics

### Why you need them
Imagine writing a long report and saving files as `report_final.docx`, `report_final2.docx`, `report_REAL_final.docx`. That gets messy fast.

- **Git** is a tool on your computer that **tracks every change** you make to your files. You can go back to any earlier version.
- **GitHub** is a website that **stores your Git projects online**. It lets you back up your work, share it, and work with other people. It is also your public portfolio for employers.

Simple way to remember: **Git is the tool, GitHub is the home online.**

### Visual: how Git and GitHub fit together

````python

fig, ax = plt.subplots(figsize=(10, 3.4))
ax.axis("off")
def box(x, y, w, h, title, sub, color):
    ax.add_patch(patches.FancyBboxPatch((x, y), w, h, boxstyle="round,pad=0.06", fc=color, ec="none"))
    ax.text(x + w/2, y + h*0.62, title, ha="center", va="center", color="white", fontsize=12, fontweight="bold")
    ax.text(x + w/2, y + h*0.28, sub, ha="center", va="center", color="white", fontsize=8.5)
box(0, 0.8, 2.3, 1.4, "Working folder", "your files", "#5B7DB1")
box(3.2, 0.8, 2.3, 1.4, "Staging area", "files ready to save", NAVY)
box(6.4, 0.8, 2.3, 1.4, "Local repository", "saved history", NAVY)
box(9.6, 0.8, 2.3, 1.4, "GitHub", "online copy", ORANGE)
for x1, x2, lab in [(2.35, 3.15, "git add"), (5.55, 6.35, "git commit"), (8.75, 9.55, "git push")]:
    ax.annotate("", xy=(x2, 1.5), xytext=(x1, 1.5), arrowprops=dict(arrowstyle="->", lw=2, color="#333"))
    ax.text((x1 + x2)/2, 2.45, lab, ha="center", fontsize=9, family="monospace", color="#333")
ax.annotate("", xy=(7.55, 0.55), xytext=(10.75, 0.55), arrowprops=dict(arrowstyle="->", lw=2, color=ORANGE, connectionstyle="arc3,rad=0"))
ax.text(9.15, 0.2, "git pull", ha="center", fontsize=9, family="monospace", color=ORANGE)
ax.set_xlim(-0.3, 12.2); ax.set_ylim(0, 3)
ax.set_title("The Git workflow: from your folder to GitHub")
plt.show()
````

### Key words
| Word | Meaning |
|---|---|
| **Repository (repo)** | A project folder that Git is tracking |
| **Commit** | A saved snapshot of your work, with a short message |
| **Branch** | A separate line of work, so you can try ideas safely |
| **Remote** | The online copy of your repo (on GitHub) |
| **Clone** | Download a copy of a GitHub repo to your computer |
| **Push** | Send your commits to GitHub |
| **Pull** | Bring new changes from GitHub to your computer |
| **Merge** | Combine one branch into another |
| **Pull request (PR)** | A request to add your branch changes into the main project |

### Step 1: Install and set up Git

Download Git from **git-scm.com** and install it. Then tell Git who you are (do this once). Type these in your terminal (Git Bash on Windows, Terminal on Mac or Linux):

```bash
git --version
git config --global user.name "Your Name"
git config --global user.email "you@example.com"
```

Use the same email as your GitHub account.

### Step 2: Create your first repository

```bash
mkdir my-first-project        # make a folder
cd my-first-project           # go inside it
git init                      # tell Git to start tracking this folder
```

You now have a repo. Git created a hidden folder called `.git` that stores the history.

### Step 3: The everyday cycle: edit, add, commit

```bash
git status                    # what has changed?
git add analysis.py           # stage one file
git add .                     # or stage everything
git commit -m "Add first analysis script"   # save the snapshot
git log --oneline             # see your history
```

**Good commit messages** are short and say what you did: `Add data cleaning step`, `Fix typo in chart title`. Avoid messages like `update` or `stuff`.

### Visual: a commit history is a chain of snapshots

````python

fig, ax = plt.subplots(figsize=(10, 2.4))
ax.axis("off")
msgs = ["Initial commit", "Add dataset", "Clean data", "Add charts", "Write README"]
for i, m in enumerate(msgs):
    ax.add_patch(plt.Circle((i * 2.2 + 0.6, 1.2), 0.28, fc=NAVY if i < len(msgs) - 1 else ORANGE, ec="none"))
    ax.text(i * 2.2 + 0.6, 0.55, m, ha="center", fontsize=9)
    if i < len(msgs) - 1:
        ax.annotate("", xy=(i * 2.2 + 2.3, 1.2), xytext=(i * 2.2 + 0.92, 1.2),
                    arrowprops=dict(arrowstyle="<-", lw=2, color="gray"))
ax.text(len(msgs) * 2.2 - 1.6, 1.8, "HEAD (you are here)", color=ORANGE, fontsize=9, fontweight="bold", ha="center")
ax.set_xlim(-0.2, 11.2); ax.set_ylim(0, 2.2)
ax.set_title("Each commit is a saved point you can return to")
plt.show()
````

### Step 4: Create a GitHub account and a remote repo

1. Go to **github.com** and sign up.
2. Click the **+** at the top right, then **New repository**.
3. Give it a name (for example `my-first-project`). Add a short description.
4. Choose **Public** (visible to everyone, good for a portfolio) or **Private**.
5. Click **Create repository**. GitHub then shows you the link to your repo.

### Step 5: Connect your computer to GitHub and push

```bash
git remote add origin https://github.com/YOUR-USERNAME/my-first-project.git
git branch -M main
git push -u origin main
```

GitHub asks you to sign in. Today it uses a **personal access token** or the **GitHub CLI / browser login**, not your normal password. After the first push, you only need `git push`.

### Step 6: Get changes back, or copy someone else's repo

```bash
git pull                                   # update your copy from GitHub
git clone https://github.com/USER/REPO.git   # download a whole project
```

### Step 7: Branches (try ideas safely)

A branch is a parallel copy of your project. The `main` branch stays safe while you experiment.

```bash
git branch                      # list branches
git switch -c new-chart         # create and move to a new branch
# ...edit files, add, commit...
git switch main                 # go back to main
git merge new-chart             # bring the new work into main
```

### Visual: branches and a merge

````python

fig, ax = plt.subplots(figsize=(10, 3))
ax.axis("off")
main_x = [0.5, 2, 3.5, 5, 6.5, 8, 9.5]
for i in range(len(main_x) - 1):
    ax.plot([main_x[i], main_x[i + 1]], [2, 2], color=NAVY, lw=3, zorder=1)
ax.scatter(main_x, [2] * len(main_x), s=240, color=NAVY, zorder=2)
# feature branch
bx = [3.5, 5, 6.5]
ax.plot([3.5, 4.2, 5, 6.5, 7.3, 8], [2, 0.9, 0.9, 0.9, 0.9, 2], color=ORANGE, lw=3, zorder=1, ls="--")
ax.scatter([5, 6.5], [0.9, 0.9], s=240, color=ORANGE, zorder=2)
ax.text(0.5, 2.5, "main", color=NAVY, fontweight="bold", ha="center")
ax.text(5.75, 0.4, "feature branch: new-chart", color=ORANGE, fontweight="bold", ha="center")
ax.text(8, 2.5, "merge", color="gray", ha="center", fontsize=10)
ax.set_xlim(0, 10.2); ax.set_ylim(0, 3)
ax.set_title("A branch splits off, then merges back into main")
plt.show()
````

### Step 8: Pull requests (how teams review work)

On GitHub, a **pull request** says: "Please review my branch and add it to the project."

1. Push your branch: `git push -u origin new-chart`
2. On GitHub, click **Compare and pull request**.
3. Write what you changed and why.
4. A teammate reviews it and clicks **Merge**.

A `README.md` is the front page of your repo. It should say:

### Git cheat sheet

| What you want | Command |
|---|---|
| Start tracking a folder | `git init` |
| See what changed | `git status` |
| Stage a file or everything | `git add file.py` or `git add .` |
| Save a snapshot | `git commit -m "message"` |
| See history | `git log --oneline` |
| Link to GitHub | `git remote add origin URL` |
| Send to GitHub | `git push` |
| Get updates | `git pull` |
| Copy a repo | `git clone URL` |
| Make a branch | `git switch -c name` |
| Change branch | `git switch name` |
| Merge a branch | `git merge name` |
| Undo changes to a file (not yet staged) | `git restore file.py` |

**Common problems**
- *"fatal: not a git repository"*: you are in the wrong folder. Use `cd` to go to your project, or run `git init`.
- *"rejected, failed to push"*: GitHub has changes you do not have. Run `git pull` first, then push again.
- *Not sure what is going on?* Run `git status`. It almost always tells you the next step.

**Using Git with Jupyter notebooks:** save the notebook, then `git add notebook.ipynb`, commit, and push. GitHub shows your notebook with all outputs, which makes a great portfolio piece.

### Quick summary

| Topic | Remember this |
|---|---|
| `print()` | Shows output on the screen |
| Variables | Names that store values. Types: int, float, str, bool |
| `type()` | Tells you the data type |
| Arithmetic | `+ - * / // % **` |
| Comparison | `== != > < >= <=` give True or False |
| Logical | `and`, `or`, `not` |
| Type change | `int()`, `float()`, `str()`. `int()` cuts off decimals |
| Strings | Join with `+`, repeat with `*`, index `[0]`, slice `[0:3]` |
| String methods | `.upper()`, `.lower()`, `.replace()`, `in` |
| Conditionals | `if`, `elif`, `else` make decisions |
| Lists | `[ ]`. Use `append`, `remove`, `extend`, `del`, `len` |
| Git | Tracks your changes on your computer |
| GitHub | Stores and shares your projects online |

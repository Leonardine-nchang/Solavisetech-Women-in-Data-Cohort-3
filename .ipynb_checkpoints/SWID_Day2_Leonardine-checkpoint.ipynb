# Part 1: Conditionals
A conditional lets your code **make a decision**. Python checks a question that has a `True` or `False` answer, then runs the right block of code.

**Remember**
- End the `if`, `elif` and `else` lines with a colon `:`
- Indent the code inside each block with four spaces

## 1.1 Variables for our examples
Here we save a temperature. The last value we save is the one Python uses, so right now `temperature` is 20.

```python
temperature = 28
temperature = 30
```

## 1.2 if and else
**Question:** is the temperature above 25?
- If yes, print "Buy icecream".
- If not, print "Drink tea".

Change the temperature to 30 and run the cell again. What happens?

```python
if temperature > 30:
    print("Buy icecream")
else:
    print("Drink tea")
```

## 1.3 elif: more than two choices
Use `elif` (short for "else if") when you have more than two choices.

**Grading rules**
| Score | Grade |
|-------|-------|
| 80 and above | A |
| 70 to 79 | B |
| 60 to 69 | C |
| Below 60 | Needs improvement |

Python checks from the top and stops at the first condition that is `True`. Try the scores 85, 72, 65 and 40.

```python
score = 40

if score >= 85:
    print("You have an A")
elif score >= 72 and score <= 65:
    print("You have a B")
elif score >= 65 and score <= 40:
    print("You have a C")
else:
    print("Average! You can do better")
```

## 1.4 Comparing values
**Do not mix these up!**

| Sign | Meaning |
|------|---------|
| `=` | Save a value in a variable |
| `==` | Ask: are these two values equal? |

Other comparison signs: `>` bigger than, `<` smaller than, `>=` bigger than or equal to, `<=` smaller than or equal to, `!=` not equal to.

```python
age = 18

print(age == 18)
print(age > 20)
print(age != 18)
```

# Part 2: Lists
A variable holds one thing. A **list** holds many items in order, inside square brackets `[ ]`, separated by commas.

**Remember:** Python starts counting at **0**, not 1.

| Position | 0 | 1 | 2 |
|----------|---|---|---|
| Item | mangoes | apples | oranges |

## 2.1 Creating a list and reading items
- `fruits[0]` is the first item
- `fruits[-1]` is the last item
- `fruits[2]` is the third item

```python
fruits = ["mangoes", "apples", "oranges"]
print(fruits[0])
print(fruits[-1])
print(fruits[2])
```

## 2.2 Adding an item with append
`.append()` adds one item to the **end** of the list.

```python
fruits.append("bananas")
print(fruits)
```

## 2.3 Removing an item with remove
`.remove()` takes out the item whose name you give it.

```python
fruits.remove("apples")
print(fruits)
```

## 2.4 Changing an item
Use the position of the item to replace it. Here we replace the first item.

```python
fruits[0] = "berries"
print(fruits)
```

## 2.5 Counting items with len
`len()` tells you how many items are in the list. Watch how the number changes after we add something.

```python
print(len(fruits))
```

```python
fruits.append("watermelon")
print(fruits)
print(len(fruits))
```

## 2.6 Adding many items with extend
`.append()` adds one item. `.extend()` adds **many items at once** from another list.

```python
new_fruits = ["cherry", "date", "elderberry"]
fruits.extend(new_fruits)
print(fruits)
```

## 2.7 Deleting items with del
`del` removes items by position. The slice `[1:3]` means "start at position 1 and stop **before** position 3". So it deletes positions 1 and 2.

```python
del fruits[1:3]
print(fruits)
```

### Quick summary
| What you want | Code |
|---------------|------|
| Read the first item | `fruits[0]` |
| Read the last item | `fruits[-1]` |
| Add one item | `fruits.append("item")` |
| Add many items | `fruits.extend(other_list)` |
| Remove by name | `fruits.remove("item")` |
| Remove by position | `del fruits[0]` |
| Change an item | `fruits[0] = "new item"` |
| Count the items | `len(fruits)` |
